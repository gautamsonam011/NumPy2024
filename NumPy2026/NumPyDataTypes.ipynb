# Data Types in NumPy

# Checking the Data Type of an Array

```python
import numpy as np

arr = np.array([1, 2, 3, 4])

print(arr.dtype)
```

```python
arrs = np.array(['apple', 'banana', 'cherry'])

print(arrs.dtype)
```

```python
arr1 = np.array([1, 2, 3, 4], dtype='S')

print(arr1)
print(arr1.dtype)
```

```python
arri = np.array([1, 2, 3, 4], dtype='i4')

print(arri)
print(arri.dtype)
```

```python
arrc = np.array(['a', '2', '3'], dtype='i')
print(arrc)
```

# Converting Data Type on Existing Arrays

```python
arras = np.array([1.1, 2.1, 3.1])

newarr = arras.astype('i')

print(newarr)
print(newarr.dtype)
```

```python
newarr = arr.astype(bool)

print(newarr)
print(newarr.dtype)
```

```python
newarr = arr.astype(int)

print(newarr)
print(newarr.dtype)
```
